```python
import shutil, zipfile, os

# Copy model to local disk
shutil.copy(
    "/content/drive/MyDrive/Toronto Pothole Training Data/Yolov8-fintuned-on-potholes.pt",
    "/content/Yolov8-fintuned-on-potholes.pt"
)
print("Model copied ✅")

# Copy zip to local disk first (avoids Drive connection drops)
shutil.copy(
    "/content/drive/MyDrive/Toronto Pothole Training Data/toronto_pothole_training.ipynb.zip",
    "/content/dataset.zip"
)
print("Zip copied ✅")

# Now unzip from local disk
with zipfile.ZipFile("/content/dataset.zip", 'r') as z:
    z.extractall("/content/dataset")

print("Dataset:", os.listdir("/content/dataset"))
```

```python
import yaml

yaml_path = "/content/dataset/data.yaml"

with open(yaml_path, 'r') as f:
    config = yaml.safe_load(f)

print("Before:", config)

config['train'] = '/content/dataset/train/images'
config['val']   = '/content/dataset/valid/images'
config['test']  = '/content/dataset/test/images'

with open(yaml_path, 'w') as f:
    yaml.dump(config, f)

print("After:", config)
```

```python
from ultralytics import YOLO

model = YOLO("/content/Yolov8-fintuned-on-potholes.pt")

results = model.train(
    data="/content/dataset/data.yaml",
    epochs=50,
    imgsz=640,
    batch=16,
    name="toronto_pothole_v3",
    project="/content/drive/MyDrive/Toronto Pothole Training Data/runs",
    patience=15,
    device=0,
    lr0=0.0001,
    lrf=0.01,
    warmup_epochs=5,
    freeze=10,
)
```
